## MatplotLib Tutorial

Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, Qt, or GTK+.

Some of the major Pros of Matplotlib are:

* Generally easy to get started for simple plots
* Support for custom labels and texts
* Great control of every element in a figure
* High-quality output in many formats
* Very customizable in general

```python
# Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
plt.title("sine wave form")

# Plot the points using matplotlib
plt.plot(x, y)
plt.show()
```

```python
#Subplot()
# Compute the x and y coordinates for points on sine and cosine curves
x = np.arange(0, 5 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Set up a subplot grid that has height 2 and width 1,
# and set the first such subplot as active.
plt.subplot(2, 1, 1)

# Make the first plot
plt.plot(x, y_sin,'r--')
plt.title('Sine')

# Set the second subplot as active, and make the second plot.
plt.subplot(2, 1, 2)
plt.plot(x, y_cos,'g--')
plt.title('Cosine')

# Show the figure.
plt.show()
```

## **Seaborn**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_flights = pd.read_csv('/content/flights.csv')
print("Dataset 'flights.csv' loaded successfully.")
```

```python
# 1. Histogram of Departure Delay
plt.figure(figsize=(10, 6))
sns.histplot(df_flights['dep_delay'], bins=50, kde=True)
plt.title('Distribution of Departure Delays')
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Frequency')
plt.show()
```

```python
# 2. Scatter Plot: Departure Delay vs. Arrival Delay
plt.figure(figsize=(10, 6))
sns.scatterplot(x='dep_delay', y='arr_delay', data=df_flights, alpha=0.5)
plt.title('Departure Delay vs. Arrival Delay')
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Arrival Delay (minutes)')
plt.show()
```

```python
# 3. Box Plot: Departure Delay by Month
plt.figure(figsize=(12, 7))
sns.boxplot(x='month', y='dep_delay', data=df_flights)
plt.title('Departure Delay by Month')
plt.xlabel('Month')
plt.ylabel('Departure Delay (minutes)')
plt.show()
```

```python
# 4. Heatmap: Flight Counts by Day of Week and Month
# First, ensure 'date' and 'day_of_week' columns are available
df_flights['date'] = pd.to_datetime(df_flights[['year', 'month', 'day']])
df_flights['day_of_week'] = df_flights['date'].dt.dayofweek # Monday=0, Sunday=6

flight_counts = df_flights.groupby(['month', 'day_of_week']).size().unstack(fill_value=0)
plt.figure(figsize=(12, 7))
sns.heatmap(flight_counts, cmap='viridis', annot=True, fmt='d', linewidths=.5)
plt.title('Flight Counts by Month and Day of Week')
plt.xlabel('Day of Week (0=Mon, 6=Sun)')
plt.ylabel('Month')
plt.show()
```

```python
# 5. Pair Plot for selected numerical features
selected_features = ['air_time', 'distance', 'dep_delay', 'arr_delay']
sns.pairplot(df_flights[selected_features].dropna())
plt.suptitle('Pair Plot of Flight Features', y=1.02) # Adjust suptitle position
plt.show()
```

## **Plotly**

```python
import plotly.express as px
import pandas as pd

# Ensure df_flights is available and has 'date' column for later plots if it wasn't re-loaded.
# This check is for robustness if previous cells were reset or not run in order.
if 'df_flights' not in locals():
    df_flights = pd.read_csv('/content/flights.csv')

# 1. Histogram of Departure Delay
fig = px.histogram(df_flights, x='dep_delay', nbins=50, title='Distribution of Departure Delays (Plotly)')
fig.update_layout(xaxis_title='Departure Delay (minutes)', yaxis_title='Frequency')
fig.show()
```

```python
# 2. Scatter Plot: Distance vs. Air Time
fig = px.scatter(df_flights, x='distance', y='air_time', title='Distance vs. Air Time (Plotly)',
                 labels={'distance': 'Distance (miles)', 'air_time': 'Air Time (minutes)'})
fig.show()
```

```python
# 3. Monthly Average Arrival Delay (Line Plot)
# Ensure 'date' column is datetime for grouping, and 'arr_delay' is ready
if 'date' not in df_flights.columns or not pd.api.types.is_datetime64_any_dtype(df_flights['date']):
    df_flights['date'] = pd.to_datetime(df_flights[['year', 'month', 'day']])

monthly_avg_arr_delay = df_flights.groupby('month')['arr_delay'].mean().reset_index()
fig = px.line(monthly_avg_arr_delay, x='month', y='arr_delay', title='Monthly Average Arrival Delay (Plotly)')
fig.update_layout(xaxis_title='Month', yaxis_title='Average Arrival Delay (minutes)')
fig.show()
```

```python
# 4. Carrier-wise Departure Delay (Bar Chart)
carrier_avg_dep_delay = df_flights.groupby('carrier')['dep_delay'].mean().reset_index()
fig = px.bar(carrier_avg_dep_delay.sort_values('dep_delay', ascending=False),
             x='carrier', y='dep_delay', title='Average Departure Delay by Carrier (Plotly)')
fig.update_layout(xaxis_title='Carrier', yaxis_title='Average Departure Delay (minutes)')
fig.show()
```

```python
# 5. Origin Airport Flight Counts (Treemap)
origin_counts = df_flights['origin'].value_counts().reset_index()
origin_counts.columns = ['origin', 'count']
fig = px.treemap(origin_counts, path=['origin'], values='count', title='Flight Counts by Origin Airport (Plotly)')
fig.show()
```
